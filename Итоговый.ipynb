```python
3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы.
```

```python
import pandas as pd
import numpy as np
import re


def experience_to_months(value):
    # Если исходное значение — пропуск
    if pd.isna(value):
        return np.nan

    # Скрытый пропуск
    if value == "Не указано":
        return np.nan

    # Ищем количество лет
    years = re.search(r'(\d+)\s*лет', value)

    # Ищем количество месяцев
    months = re.search(r'(\d+)\s*месяц', value)

    years_count = int(years.group(1)) if years else 0
    months_count = int(months.group(1)) if months else 0

    # Переводим всё в месяцы
    return years_count * 12 + months_count


project_df["Опыт работы (месяц)"] = project_df["Опыт работы"].apply(experience_to_months)

# Удаляем исходный столбец

project_df.head()
```

Создадим отдельные признаки **"Город"**, **"Готовность к переезду"**, **"Готовность к командировкам"**. При этом важно учесть:

* Признак **"Город"** должен содержать только 4 категории: "Москва", "Санкт-Петербург" и "город-миллионник" (их список ниже), остальные обозначьте как "другие".

   <code>million_cities = ['Новосибирск', 'Екатеринбург','Нижний Новгород','Казань', 'Челябинск','Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж','Волгоград']
    </code>
    Инфорация о метро, рядом с которым проживает соискатель нас не интересует.
* Признак **"Готовность к переезду"** должен иметь два возможных варианта: True или False. Обратите внимание, что возможны несколько вариантов описания готовности к переезду в признаке "Город, переезд, командировки". Например:
    * … , готов к переезду , …
    * … , не готова к переезду , …
    * … , готова к переезду (Москва, Санкт-Петербург, Ростов-на-Дону)
    * … , хочу переехать (США) , …
    
    Нас интересует только сам факт возможности или желания переезда.
* Признак **"Готовность к командировкам"** должен иметь два возможных варианта: True или False. Обратите внимание, что возможны несколько вариантов описания готовности к командировкам в признаке "Город, переезд, командировки". Например:
    * … , готов к командировкам , … 
    * … , готова к редким командировкам , …
    * … , не готов к командировкам , …
    
    Нас интересует только сам факт готовности к командировке.
    
    Еще один важный факт: при выгрузки данных у некоторых соискателей "потерялась" информация о готовности к командировкам. Давайте по умолчанию будем считать, что такие соискатели не готовы к командировкам.

```python
million_cities = ['Новосибирск', 'Екатеринбург','Нижний Новгород','Казань', 
                  'Челябинск','Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 
                  'Красноярск', 'Пермь', 'Воронеж','Волгоград']

# Город
def extract_city(text):
    if pd.isna(text):
        return None
    parts = text.split(' , ')
    city_raw = parts[0]
    if '(' in city_raw:
        city_raw = city_raw.split('(')[0].strip()
    return city_raw

# Категория города
def get_city_category(city):
    if city == 'Москва':
        return 'Москва'
    elif city == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'

# Готовность к переезду
def extract_relocation(text):
    if pd.isna(text):
        return False
    if 'не готов к переезду' in text or 'не готова к переезду' in text:
        return False
    elif 'готов к переезду' in text or 'готова к переезду' in text:
        return True
    elif 'хочу переехать' in text:
        return True
    else:
        return False

# Готовность к командировкам
def extract_business_trips(text):
    if pd.isna(text):
        return False
    if 'командировкам' not in text:
        return False
    if 'не готов к командировкам' in text or 'не готова к командировкам' in text:
        return False
    elif 'готов к командировкам' in text or 'готова к командировкам' in text:
        return True
    elif 'редким командировкам' in text:
        return True
    else:
        return False
# Создаём новые столбцы
project_df['Город'] = project_df['Город, переезд, командировки'].apply(extract_city)
project_df['Категория города'] = project_df['Город'].apply(get_city_category)
project_df['Готовность к переезду'] = project_df['Город, переезд, командировки'].apply(extract_relocation)
project_df['Готовность к командировкам'] = project_df['Город, переезд, командировки'].apply(extract_business_trips)

project_df.head()
```

Осталось только понять, откуда брать дату, по которой определяется курс? А вот же она - в признаке **"Обновление резюме"**, в нем содержится дата и время, когда соискатель выложил текущий вариант своего резюме. Нас интересует только дата, по ней бы и будем сопоставлять курсы валют.

Теперь у нас есть вся необходимая информация для того, чтобы создать признак "ЗП (руб)" - заработная плата в рублях.

```python
exchange_rates = pd.read_csv(r"C:\Users\nzena\Downloads\ExchangeRates.csv")

# Даты
project_df['Обновление резюме'] = pd.to_datetime(
    project_df['Обновление резюме'],
    dayfirst=True
)

exchange_rates['date'] = pd.to_datetime(
    exchange_rates['date'],
    dayfirst=True
)

project_df['date'] = project_df['Обновление резюме'].dt.normalize()

# Сумма зарплаты
project_df['ЗП сумма'] = (
    project_df['ЗП']
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

# Валюта
project_df['currency'] = project_df['ЗП'].str.extract(
    r'\d+\s*(\S+)'
)[0]

# Соответствие валют
currency_map = {
    'руб.': 'RUB',
    'руб': 'RUB',
    'грн.': 'UAH',
    'грн': 'UAH',
    'USD': 'USD',
    'EUR': 'EUR',
    'белруб': 'BYN',
    'KGS': 'KGS',
    'сум': 'UZS',
    'AZN': 'AZN',
    'KZT': 'KZT'
}

project_df['currency'] = project_df['currency'].map(currency_map)

# Объединяем по дате и валюте
project_df = project_df.merge(
    exchange_rates[['currency', 'date', 'close', 'proportion']],
    how='left',
    on=['currency', 'date']
)

# Для рублей курс равен 1
project_df.loc[
    project_df['currency'] == 'RUB',
    ['close', 'proportion']
] = 1

# Расчёт зарплаты в рублях
project_df['ЗП (руб)'] = (
    project_df['ЗП сумма'] *
    project_df['close'] /
    project_df['proportion']
)

# # Удаляем исходный и промежуточные столбцы
# project_df.drop(
#     columns=[
#         'ЗП',
#         'ЗП сумма',
#         'currency',
#         'date',
#         'close',
#         'proportion'
#     ],
#     inplace=True
# )

project_df
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Логарифмируем возраст
log_age = np.log(project_df['Возраст'])

# Считаем среднее и стандартное отклонение
mean_log = log_age.mean()
std_log = log_age.std()

# Границы трех сигм (с послаблением +4 сигмы справа)
lower = mean_log - 3 * std_log
upper = mean_log + 4 * std_log

# Строим график
plt.figure(figsize=(10, 6))
sns.histplot(log_age, bins=30, kde=True, color='skyblue')

plt.axvline(mean_log, color='red', linestyle='--', linewidth=2, label='Среднее')
plt.axvline(lower, color='green', linestyle='--', linewidth=2, label='-3σ')
plt.axvline(upper, color='orange', linestyle='--', linewidth=2, label='+4σ')

plt.title('Распределение логарифма возраста')
plt.xlabel('log(Возраст)')
plt.ylabel('Частота')
plt.legend()
plt.show()



```

Распределение асимметрично вправо: есть выбросы с аномально большим возрастом.

```python
# Z-отклонение
z_scores = (log_age - mean_log) / std_log

# Выбросы — те, у кого z > 4
outliers = project_df[z_scores > 4]

# Смотрим, что за возраст у выбросов
print(f"Найдено выбросов: {len(outliers)}")
print(outliers[['Возраст', 'ЗП (руб)']].sort_values('Возраст'))

# Удаляем их
project_df = project_df[z_scores <= 4]

print(f"Осталось записей: {len(project_df)}")
```
